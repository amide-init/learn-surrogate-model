## Part 1 — NumPy Basics

NumPy lets you work with arrays of numbers instead of Python lists.
The key advantage: you can apply math to every element at once — no loops needed.

````python
# Create an array of 5 evenly spaced numbers between 0 and 1
a = np.linspace(0, 1, 5)
print('Array:', a)
print('Shape:', a.shape)
print('Data type:', a.dtype)
````

````python
# Vectorised math — no for loop needed
print('a * 2  :', a * 2)
print('a ** 2 :', a ** 2)
print('sin(a) :', np.sin(a))
````

````python
# 2D arrays (matrices)
M = np.array([[1, 2], [3, 4]])
print('Matrix M:')
print(M)
print('Shape:', M.shape)
print('Transpose:')
print(M.T)
````

## Part 2 — Plotting with Matplotlib

Every plot must have a title, axis labels, and a legend.  
Always follow this pattern:

```python
fig, ax = plt.subplots()
ax.plot(x, y, label='...')
ax.set_title('...')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
plt.show()
```

````python
# Plot sin(x) and x² on the same figure
x = np.linspace(-2 * np.pi, 2 * np.pi, 300)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(x, np.sin(x), label='sin(x)', color='steelblue')
axes[0].axhline(0, color='black', linewidth=0.5)
axes[0].set_title('Sine function')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(x, x**2, label='x²', color='tomato')
axes[1].set_title('Quadratic function')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Part 3 — 2D Contour Plot

In optimisation, we often work with functions that take two inputs: `f(x, y)`.  
A contour plot shows lines of equal function value — like a topographic map.  

We will plot `f(x, y) = x² + y²` — a simple bowl shape with its minimum at (0, 0).

````python
# Build a grid of (x, y) points
grid_x = np.linspace(-3, 3, 200)
grid_y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(grid_x, grid_y)

print('X shape:', X.shape)  # 200 x 200
print('Y shape:', Y.shape)
````

````python
# Evaluate f(x, y) = x² + y² at every grid point
Z = X**2 + Y**2

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Filled contour plot
cf = axes[0].contourf(X, Y, Z, levels=20, cmap='viridis')
axes[0].contour(X, Y, Z, levels=20, colors='white', linewidths=0.4, alpha=0.5)
fig.colorbar(cf, ax=axes[0], label='f(x, y)')
axes[0].plot(0, 0, 'r*', markersize=12, label='minimum (0, 0)')
axes[0].set_title('f(x, y) = x² + y²  —  contour')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].legend()

# 3D surface
ax3d = fig.add_subplot(1, 2, 2, projection='3d')
ax3d.plot_surface(X, Y, Z, cmap='viridis', alpha=0.85, linewidth=0)
ax3d.set_title('f(x, y) = x² + y²  —  surface')
ax3d.set_xlabel('x')
ax3d.set_ylabel('y')
ax3d.set_zlabel('f(x, y)')

plt.tight_layout()
plt.show()
````

**Try it yourself:** Replace `Z = X**2 + Y**2` with the Rosenbrock function:
```python
Z = (1 - X)**2 + 100 * (Y - X**2)**2
```
Notice how the valley curves — this function is much harder to optimise.

## Part 4 — Finding a Minimum with SciPy

`scipy.optimize.minimize` takes a function and a starting point, and tries to find the minimum.

We will minimise `f(x) = (x - 2)² + 1`.  
The true minimum is at `x = 2`, where `f(2) = 1`.

````python
def f(x):
    # scipy passes x as an array, so we use x[0]
    return (x[0] - 2) ** 2 + 1

x0 = [0.0]  # starting point
result = minimize(f, x0, method='L-BFGS-B')

print('Minimum at x =', result.x[0])
print('f(x)         =', result.fun)
print('Converged    =', result.success)
print('Message      =', result.message)
````

````python
# Visualise the optimisation
x_plot = np.linspace(-1, 5, 300)
y_plot = (x_plot - 2) ** 2 + 1

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x_plot, y_plot, label='f(x) = (x − 2)² + 1', color='steelblue', linewidth=2)
ax.plot(x0[0], f(x0), 'go', markersize=10, label=f'start  x₀ = {x0[0]}')
ax.plot(
    result.x[0], result.fun, 'r*', markersize=14,
    label=f'minimum  x = {result.x[0]:.4f},  f = {result.fun:.4f}'
)
ax.set_title('Minimising f(x) = (x − 2)² + 1 with SciPy')
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

**Try it yourself:** Change `method='L-BFGS-B'` to `method='Nelder-Mead'`. Does it find the same minimum?

## Part 5 — NumPy Random Sampling

In later lessons we will sample random points to build our initial dataset.  
Here is how random sampling works in NumPy.

````python
rng = np.random.default_rng(seed=42)

# 10 random points in [0, 1]
points_1d = rng.uniform(0, 1, size=10)
print('1D random points:', points_1d)

# 10 random points in 2D space [0,1]²
points_2d = rng.uniform(0, 1, size=(10, 2))
print('\n2D random points (shape:', points_2d.shape, '):')
print(points_2d)
````

````python
# Plot the 2D points
fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(points_2d[:, 0], points_2d[:, 1], color='steelblue', s=60, zorder=3)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_title('10 random points in [0, 1]²')
ax.set_xlabel('x₁')
ax.set_ylabel('x₂')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````
